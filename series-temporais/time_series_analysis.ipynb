```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
url = 'https://raw.githubusercontent.com/jbrownlee/Datasets/refs/heads/master/airline-passengers.csv'

df = pd.read_csv(url, header = 0, index_col = 0,parse_dates=True)
```

```python
#Decomposição de Séries Temporais

result_mul = seasonal_decompose(df, model='multiplicativo')
result_add = seasonal_decompose(df, model='Aditivo')
```

```python
#Corrija o subplots para que eles não se sobreponham
plt.rcParams.update({'figure.figsize': (8,5)})
result_mul.plot()
plt.show()
```

```python
plt.rcParams.update({'figure.figsize': (8,5)})
result_add.plot()
plt.show()
```
