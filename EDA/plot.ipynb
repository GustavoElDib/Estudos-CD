```python
base_co2 = pd.read_csv('co2.csv')
base_co2.head()
```

```python
x = base_co2.conc
y = base_co2.uptake
```

```python
categorica = list(set(base_co2.Treatment))
categorica
```

```python
for i in range(len(categorica)):
    indice = base_co2.Treatment == categorica[i]
    plt.scatter(x[indice], y[indice], label = categorica[i])
plt.legend(loc = 'lower right')
```
